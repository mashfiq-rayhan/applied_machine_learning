##### Iris Flower Dataset

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

iris = pd.read_csv('./Materials/Iris.csv')
iris
```

```python
# Q. how many data points and features?
iris.shape
```

```python
# what are the column names in our dataset?
iris.columns
```

```python
# Q. How many data points for each class are present?
# Q. How many flowers for each species are present?
iris['species'].value_counts()
# balanced dataset vs imbalanced datasets
# Iris is a balanced dataset
```

#### 2D Scatter Plot

```python
# 2D Scatter Plot
# Always Understand the axis: labels and scale
# plt.style.use("dark_background")
iris.plot(kind='scatter', x='sepal_length', y='sepal_width')
plt.show()
```

```python
# 2D Scatter plot with color-coding for each flower type/class/species
sns.set_style("whitegrid")
g = (
    sns.FacetGrid(iris, hue="species", height=4, palette="bright")
    .map(plt.scatter, "sepal_length", "sepal_width")
    .add_legend()
)

plt.show()
```
