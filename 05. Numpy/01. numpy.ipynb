```python
x = np.array([1, 2, 3, 1])
y = np.array([[1, 2, 3], [5, 6, 1]])

print(x.mean())     # Mean: the average value of all elements in x
print(np.median(x))     # Median: the middle value when the elements of x are sorted
print(np.median(y, axis=-1))
print(x.std())      # Standard Deviation: how spread out the values are from the mean.
```
